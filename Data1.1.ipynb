```python
# I import library : numpy / pandas / matplotlib / seaborn

import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 

# it set_option show a full columns and full rows
pd.set_option('display.max_info_columns',100)
pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
```

```python
df = df.apply(lambda x: x.astype(str).str.lower().str.strip().str.replace(' ','_')if x.dtype == 'object' else x )
df.columns = df.columns.str.lower().str.strip().str.replace(' ','_')
df.head()
```

```python
df = df.replace({'yes':1 , 'no':0})
df = df.rename(columns={'rapidcharge':'rapidcharge_y1_n0'})
df.head()
```

```python
df_brand = df.brand.value_counts().head(20)

plt.figure(figsize=(6,4))
sns.barplot(x=df_brand.values,y=df_brand.index,palette='husl')
plt.grid(True)
plt.gca().set_axisbelow(True)
plt.show()

```

```python
df.head()
```

```python
plt.figure(figsize=(6,6))
plt.scatter(x=df.powertrain,y=df.topspeed_kmh,c='g',alpha=0.6)
plt.grid(True)
plt.gca().set_axisbelow(True)
plt.xlabel('PowerTrain')
plt.ylabel('Topspeed_kmh')
plt.title('Power_Train VS Top_Speed')
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df, x='range_km', y='priceeuro',hue='bodystyle',size='topspeed_kmh',sizes=(20,200)
)

sns.regplot(data=df, x='range_km',y='priceeuro',scatter=False,color='black',line_kws={'linestyle': '--', 'linewidth': 2})

plt.legend(bbox_to_anchor=(1.05,1) ,loc='upper left')
plt.grid(True)
plt.gca().set_axisbelow(True)
plt.show()

```
