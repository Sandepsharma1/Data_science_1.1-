```python
df = df.apply(lambda x : x.astype(str).str.lower().str.strip().str.replace(' ','_')if x.dtype == 'object' else x)
df.columns =  df.columns.str.lower().str.strip().str.replace(' ','_')


df_lower = ['school','sex','address','famsize','pstatus']
df[df_lower] = df[df_lower].apply(lambda x :x.str.lower())

df.head()
```

```python
df['sex'] = df['sex'].map({'f' : 1 , 'm' : 0}).astype(int)


df.rename(columns={'sex':'sex_f1_m0'},inplace=True)

df.replace({'yes':1,'no':0},inplace=True)

df.head()
```

```python
df_data_frame = ['school','age','sex_f1_m0','address','mjob','fjob']
for col in df_data_frame:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col],kde=True,bins=20)
```

```python
df = pd.get_dummies(df,columns=['reason'],dtype=int)
df.head()
```

```python
df = pd.get_dummies(df,columns=['guardian'],dtype=int)
df.head()
```
